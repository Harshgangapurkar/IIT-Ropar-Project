#🎵 Song Recommendation System (Content-Based)

## Project Overview
This project implements a **content-based song recommendation system** that suggests songs similar to a given input song.

The system analyzes **textual information such as lyrics and song metadata** and recommends songs using similarity-based techniques without relying on user listening history.


## 1. Problem Definition & Objective

### a. Selected Project Track
This project falls under the **Recommendation Systems** track, combining concepts from Machine Learning and Natural Language Processing (NLP).

### b. Problem Statement
Music streaming platforms host millions of songs, making it difficult for users to discover music aligned with their preferences.

The objective of this project is to build a **song recommendation system** that recommends similar songs based on content features derived from song lyrics and metadata.

### c. Real-World Relevance and Motivation
Content-based song recommenders are widely used in platforms like Spotify and Apple Music, particularly in:
- Cold-start scenarios
- Discovering similar songs without user history
- Transparent and explainable recommendations


# libraries

```python
import pandas as pd
import nltk
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
from nltk.stem.porter import PorterStemmer
nltk.download("punkt")
nltk.download("punkt_tab")
```

# 2. Data Understanding & Preparation

### a. Dataset Source
The dataset used in this project is a **publicly available song dataset** containing song titles, artists, and textual content such as lyrics.

The dataset is suitable for building a text-based recommendation system.

### b. Data Loading and Exploration
The dataset is loaded using standard data processing libraries.  
Initial exploration is performed to understand:
- Number of songs
- Available features
- Textual structure of the data


```python
url = "https://drive.google.com/uc?export=download&id=12W8B6Q2SKXawlo28pMqJk9t_QRcnpn1D"
df = pd.read_csv(url)
df = df.drop(columns=["link"], errors="ignore")
```

### c. Data Cleaning, Preprocessing, and Feature Engineering
To prepare song data for similarity computation:
- Text is converted to lowercase
- Noise such as special characters is removed
- Text is tokenized and normalized
- Stemming is applied to reduce words to their root forms

These steps improve the quality of text representations used for similarity analysis.

### d. Handling Missing Values or Noise
Rows with missing or insufficient textual information are handled appropriately to avoid incorrect similarity computation.


#3. Model / System Design

### a. AI Technique Used
This project uses a **content-based recommendation approach** combined with **Natural Language Processing (NLP)** techniques.

TF-IDF vectorization is used to convert song lyrics into numerical representations.

### b. System Pipeline
The recommendation pipeline consists of:
1. Text preprocessing of song lyrics
2. TF-IDF vectorization
3. Similarity computation using cosine similarity
4. Ranking and selection of similar songs

### c. Justification of Design Choices
TF-IDF was selected because it:
- Captures important words in lyrics
- Handles sparse text efficiently
- Works well with cosine similarity for recommendation tasks


# Cleaning text

```python
df["text"] = df["text"].str.lower().replace(r"\n", " ", regex=True)

stemmer = PorterStemmer()
def stem_text(text):
    return " ".join(stemmer.stem(w) for w in nltk.word_tokenize(text))

df["text"] = df["text"].apply(stem_text)
```

# 4. Core Implementation

### a. Model Training / Inference Logic
This system does not use supervised learning.  
Instead, it performs inference by comparing TF-IDF vectors of songs using cosine similarity.

### b. Recommendation Pipeline
Given an input song:
- The song is vectorized
- Similarity scores are computed against all other songs
- Songs are ranked based on similarity
- The top-N most similar songs are recommended

### c. Execution Consistency
The notebook is designed to run sequentially from top to bottom without errors, ensuring reproducibility.


# TF-IDF

```python
tfidf = TfidfVectorizer(stop_words="english", max_features=10000)
tfidf_matrix = tfidf.fit_transform(df["text"])

# Nearest Neighbors (NO COSINE MATRIX)
nn_model = NearestNeighbors(metric="cosine", algorithm="brute", n_neighbors=6)
nn_model.fit(tfidf_matrix)
```

#5. Evaluation & Analysis

### a. Evaluation Metrics Used
Since this is a recommendation system without labeled relevance data, traditional classification metrics are not applicable.

The following evaluation strategies are used:
- Self-similarity validation

- Qualitative inspection of recommendations

### b. Quantitative Results
Similarity scores indicate that recommended songs are closely related to the input song in terms of lyrical content.

### c. Qualitative Analysis
Manual inspection shows that recommended songs often share:

### d. Limitations of Evaluation
